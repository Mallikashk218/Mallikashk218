**Univariate Analysis**

Univariate analysis is the process of analyzing a single variable to understand its distribution, frequency, and patterns using charts or statistical measures.

 **Lead time using histogram**

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(8, 5))
plt.hist(data['lead_time'], bins=20, color='skyblue', edgecolor='black')
plt.title('Distribution of Lead Time')
plt.xlabel('Lead Time (days)')
plt.ylabel('Number of Bookings')
plt.show()
```

**Insight**:

* Most bookings are made close to arrival.
* The median lead time is 49 days, and about 40% of bookings are made within 30 days.  
* A few bookings are made nearly two years ahead (maximum 709 days), so the chart has a long tail to the right.

**Average daily rate, ADR using Histogram**

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(8, 5))
plt.hist(data['adr'], bins=30, color='orange', edgecolor='black')
plt.title('Distribution of Average Daily Rate (ADR)')
plt.xlabel('ADR')
plt.ylabel('Number of Bookings')
plt.show()
```

**Insight**:
* Most rates fall between about 73 and 135, with a median of 99.
* A few high-priced bookings stretch the right side.

**Total nights stayed using Bar chart**

```python
data['total_nights'] = data['stays_in_weekend_nights'] + data['stays_in_weekdays_nights']

plt.figure(figsize=(8, 5))
sns.countplot(x='total_nights', data=data[data['total_nights'] <= 14], color='skyblue', edgecolor='black')
plt.title('Number of Bookings by Total Nights Stayed')
plt.xlabel('Total Nights')
plt.ylabel('Number of Bookings')
plt.show()
```

**Insight**:

**Market segment (bar chart)**

**Bivariate analysis**

Bivariate analysis is the process of analyzing two variables to understand the relationship between them.

**Hotel type vs cancellation rate**

```python
rate = data.groupby('hotel')['is_canceled'].mean() * 100

rate.plot(kind='bar', color=['blue', 'orange'], edgecolor='black')
plt.title('Cancellation Rate by Hotel Type')
plt.xlabel('Hotel Type')
plt.ylabel('Cancellation Rate (%)')
plt.xticks(rotation=0)
plt.show()
```

**Insight:**  

* City Hotel is cancelled more often, at 30.3% against 23.7% for Resort Hotel.

**Total nights vs cancellation rate**

```python
data['total_nights'] = data['stays_in_weekend_nights'] + data['stays_in_weekdays_nights']
```

```python
# 1. Keep only stays of 14 nights or less
short = data[data['total_nights'] <= 14]

# 2. Work out the cancellation % for each number of nights, for each hotel
rate = short.groupby(['total_nights', 'hotel'])['is_canceled'].mean() * 100

# 3. Draw the line chart
rate.unstack().plot(marker='o', figsize=(10, 5))
plt.title('Cancellation Rate by Length of Stay')
plt.xlabel('Total Nights Stayed')
plt.ylabel('Cancellation Rate (%)')
plt.show()
```

**Insight:**

* Longer stays are cancelled more often at both hotels.
* City Hotel climbs much faster: 22% for 1 night, 34% for 5-7 nights, and 52% for 8-14 nights.



**Lead time vs cancellation rate**

```python
data['lead_group'] = pd.cut(data['lead_time'],
                            bins=[-1, 7, 30, 90, 180, 365, 1000],
                            labels=['0-7', '8-30', '31-90', '91-180', '181-365', '365+'])

rate = data.groupby(['lead_group', 'hotel'], observed=True)['is_canceled'].mean() * 100
rate.unstack().plot(kind='bar', figsize=(10, 5), edgecolor='black')
plt.title('Cancellation Rate by Lead Time')
plt.xlabel('Lead Time (days)')
plt.ylabel('Cancellation Rate (%)')
plt.xticks(rotation=0)
plt.show()
```

**Insight**:

 Last-minute bookings (0-7 days) are cancelled least, at about 11% for City Hotel and 6% for Resort Hotel. Cancellations rise steadily with lead time, reaching about 45% at City Hotel for bookings made 181-365 days ahead. Resort Hotel is more stable.

**Market segment vs cancellation rate**

```python
seg = data.groupby('market_segment')['is_canceled'].mean() * 100

seg.sort_values(ascending=False).plot(kind='bar', color='skyblue', figsize=(10, 5), edgecolor='black')
plt.title('Cancellation Rate by Market Segment')
plt.xlabel('Market Segment')
plt.ylabel('Cancellation Rate (%)')
plt.xticks(rotation=45)
plt.show()
```

 **Insight**:

* Online travel agent bookings are cancelled the most (35.5%). Direct (14.8%), Offline TA (15.0%) and Corporate (12.1%) are much lower.


**Month vs number of bookings, by hotel**

```python
month_order = ['January', 'February', 'March', 'April', 'May', 'June', 'July',
               'August', 'September', 'October', 'November', 'December']

monthly = data.groupby(['arrival_date_month', 'hotel']).size().unstack()
monthly = monthly.reindex(month_order)

monthly.plot(marker='o', figsize=(10, 5))
plt.title('Number of Bookings per Month')
plt.xlabel('Month')
plt.ylabel('Number of Bookings')
plt.xticks(range(12), month_order, rotation=45)
plt.show()
```


**Insight:**
*  Bookings are highest in September and October for both hotels (City peaks at 6,473 in October). They are lowest in January to March, with March the lowest.

**Correlation between key variables**

```python
#heatmap of correlation between key variables
plt.figure(figsize = (8,6))
cols = ['lead_time', 'total_nights', 'adr', 'total_of_special_requests', 'booking_changes', 'is_canceled']
sns.heatmap(data = data[cols].corr(), annot = True, fmt = '.2f', cmap = 'coolwarm', center = 0)
plt.title('heatmap of correlation between key variables\n', color = 'brown')
plt.show()
```

*    This heatmap relates six variables at once, which is what makes it multivariate.



**Cancellation rate by market segment and lead time**

```python
#heatmap of cancellation rate by market_segment v/s lead_group
plt.figure(figsize = (8,6))
segs = ['Online TA', 'Offline TA/TO', 'Direct', 'Corporate', 'Groups']
table = data[data['market_segment'].isin(segs)].pivot_table(values = 'is_canceled',
                                                           index = 'market_segment',
                                                           columns = 'lead_group',
                                                           aggfunc = 'mean',
                                                           observed = True) * 100
sns.heatmap(data = table, annot = True, fmt = '.0f', cmap = 'YlOrRd')
plt.title('heatmap of cancellation rate (%) by market_segment v/s lead_group\n', color = 'brown')
plt.show()
```

**Insight**:

*  Online travel agent bookings climb from 9% (0-7 days) to 52% (181-365 days), while Offline travel agent bookings only rise from 7% to 20%. Lead time has a much bigger effect on online bookings.
*  This heatmap uses three variables together: market segment, lead time and cancellation.


**Conclusion and Recommendations**

**Conclusion**

* City Hotel gets 61% of bookings against 39% for Resort Hotel. Bookings peak in September and October and are lowest in January to March.

* Longer stays are cancelled more. City Hotel is cancelled more overall (30.3% against 23.7%), and it reaches 52% for 8-14 nights and 77% for 15+ nights.

* Cancellation rises as bookings are made further ahead. It goes from 11% for last-minute bookings to 45% for 181-365 days ahead at City Hotel. Resort Hotel is more stable.

**Recommendations**

   *  Raise prices in September and October, and offer discounts and packages in January to March. Promote the Resort Hotel with weekend and festival deals.

   * Ask for a deposit or use tiered cancellation fees on City Hotel stays of 8+ nights.

   * For bookings made 6+ months ahead, send reminders, take a partial deposit, and offer free date changes instead of cancellation.

Put cancellation protection (a deposit or cancellation fee) on far-ahead and long-stay City Hotel bookings, especially those from online travel agents. City Hotel is the larger hotel and has the highest cancellation rates (45% to 77%), so this recovers the most lost revenue.
